### Scatter Plot
Number to Number

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
def days_color(days):
    if days == 'Sun':
        return 0.25
    elif days == 'Sat':
        return 0.5
    elif days == 'Fri':
        return 0.75
    else:
        return 1
```

```python
days_colors = tips['day'].apply(days_color)
```

```python
# days = tips['day'].unique()



plt.scatter(tips['total_bill'],tips['tip'], s=tips['size']*20, cmap='viridis' , c = days_colors)
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Scatter Plot')
plt.grid()
plt.legend()
plt.show()
```
