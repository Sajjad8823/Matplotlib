```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Bar Chart
#### Categorical to Numerical

```python
df_sales = pd.read_csv('sales_data.csv')
df_sales.head(5)
```

```python
categories = df_sales.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)

plt.bar(categories.index, categories.values , color='green', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
plt.xlabel('Product Category')
plt.ylabel('Sales Amount')
plt.title('Sales by Product Category')
plt.show()
```

```python
categories
```

```python
plt.barh(categories.index, categories.values , color='green', edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
plt.xlabel('Product Category')
plt.ylabel('Sales Amount')
plt.title('Sales by Product Category')
plt.show()
```

```python
region = df_sales.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)
plt.pie(region.values, labels=region.index, autopct='%1.2f%%',  explode=(0.1, 0, 0.2, 0), shadow=True)
plt.title('Sales by Region')
plt.show()
```

```python
fig , ax = plt.subplots(1,2)
ax[0].bar(categories.index, categories.values , color='green', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
ax[1].pie(region.values, labels=region.index, autopct='%1.2f%%',  explode=(0.1, 0, 0.2, 0), shadow=True)
ax[0].set_xlabel('Product Category')
ax[0].set_ylabel('Sales Amount')
ax[0].set_title('Sales by Product Category')
ax[1].set_title('Sales by Region')
ax[1].set_ylabel('Region')

plt.show()
```

```python
fig , ax = plt.subplots(2,2, figsize=(14,8), sharex=True, )
ax[0][0].bar(categories.index, categories.values , color='green', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
ax[0][1].pie(region.values, labels=region.index, autopct='%1.2f%%',  explode=(0.1, 0, 0.2, 0), shadow=True)
ax[1][0].bar(categories.index, categories.values , color='green', edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
ax[1][1].pie(region.values, labels=region.index, autopct='%1.2f%%',  explode=(0.1, 0, 0.2, 0), shadow=True)
ax[0][0].set_xlabel('Product Category')
ax[0][0].set_ylabel('Sales Amount')
ax[0][0].set_title('Sales by Product Category')
ax[0][1].set_title('Sales by Region')
ax[0][1].set_ylabel('Region')
ax[1][0].set_xlabel('Product Category')
ax[1][0].set_ylabel('Sales Amount')
ax[1][0].set_title('Sales by Product Category')
ax[1][1].set_title('Sales by Region')
ax[1][1].set_ylabel('Region')

plt.show()
```

### Histogram
#### Numerical to Numerical

```python

plt.hist(df_sales['Sales_Amount'], bins=[0, 1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000, 10000], edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.title('Distribution of Sales Amount')
plt.show()
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
plt.hist(titanic['age'])
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Age')
plt.show()
```

```python
plt.hist(titanic['age'], bins=[0, 30, 60,80], edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Age')
plt.show()
```

```python
pclass_survived = titanic.groupby(['pclass','survived'])['survived'].count()
pclass_survived
```

```python
pclass_survived.values
```

```python
plt.bar(pclass_survived.index, pclass_survived.values , color='green', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', )
plt.xlabel('Pclass')
plt.ylabel('Survived')
plt.title('Survival by Pclass')
plt.show()
```

```python
titanic.head()
```

```python
pclass_survived_pivot = pclass_survived.unstack()
pclass_survived_pivot
```

```python
plt.bar(pclass_survived_pivot.index-0.5, pclass_survived_pivot[0] , color='red', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*', label='Not Survived')
plt.bar(pclass_survived_pivot.index, pclass_survived_pivot[1] , color='green', width=0.5 , edgecolor='black', linewidth=2, alpha=0.7, hatch='x*',label='Survived')
plt.xlabel('Pclass')
plt.ylabel('Survived')
plt.title('Survival by Pclass')
plt.legend()
plt.show()
```
