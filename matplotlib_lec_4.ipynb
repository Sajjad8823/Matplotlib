```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### 3D Plots

```python
x = np.linspace(0,10,100)
y = np.linspace(0,10,100)
z = x**2+y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(x,y,z)
plt.show()
```

```python
fig = plt.figure()
ax = plt.subplot(111, projection='3d')
ax.plot(x,y,z)
plt.show()
```

```python
x = np.random.randint(0,10,100)
y = np.linspace(0,10,100)
z = np.linspace(0,10,100)

fig = plt.figure()
ax =  plt.subplot(111, projection='3d')
ax.scatter(x,y,z)
plt.show()
```

```python
# Surface Plot
x = np.linspace(-100,100,100)
y = np.linspace(-100,100,100)

X,Y = np.meshgrid(x,y)
z = X**2 + Y**2

fig = plt.figure()
ax = plt.subplot(111, projection='3d')
surface = ax.plot_surface(X,Y,z, cmap='viridis')
plt.colorbar(surface)
plt.show()
```

```python
# Surface Plot
x = np.linspace(-100,100,100)
y = np.linspace(-100,100,100)

X,Y = np.meshgrid(x,y)
z = X**2 + Y**2

fig = plt.figure()
ax = plt.subplot(111, projection='3d')
surface = ax.contourf(X,Y,z, cmap='viridis')
plt.colorbar(surface)
plt.show()
```

```python
c = plt.contourf(X,Y,z)
plt.colorbar(c)
plt.show()
```

```python
# Surface Plot
x = np.linspace(-100,100,100)
y = np.linspace(-100,100,100)

X,Y = np.meshgrid(x,y)
z = np.sin(X**2 + Y**2)

fig = plt.figure()
ax = plt.subplot(111, projection='3d')
surface = ax.plot_surface(X,Y,z, cmap='viridis')
plt.colorbar(surface)
plt.show()
```
